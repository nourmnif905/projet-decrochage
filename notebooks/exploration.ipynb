Colonnes constantes ou presque : une valeur dominante à plus de 95 % apporte peu d'information.

```python
dominante = X.apply(lambda s: s.value_counts(normalize=True).iloc[0])
dominante.sort_values(ascending=False).head(10).to_frame("part de la valeur dominante")
```

Valeurs extrêmes (outliers, méthode IQR)

```python
num_apercu = X.select_dtypes(include="number")
Q1, Q3 = num_apercu.quantile(0.25), num_apercu.quantile(0.75)
IQR = Q3 - Q1
outliers = ((num_apercu < Q1 - 1.5 * IQR) | (num_apercu > Q3 + 1.5 * IQR)).sum()
outliers.sort_values(ascending=False).head(10).to_frame("nb_outliers")
```

## E. Relation entre les variables et le décrochage

```python
df = X.copy()
df["Dropout"] = y
```

Variables à 2 valeurs (binaires)

```python
bin_cols = [c for c in X.columns if X[c].nunique() == 2]
print("Variables binaires :", bin_cols)

for col in bin_cols:
    taux = df.groupby(col)["Dropout"].mean().reset_index()
    fig = px.bar(taux, x=col, y="Dropout",
                 title=f"Taux de décrochage selon : {col}",
                 labels={"Dropout": "Taux de décrochage"})
    fig.update_xaxes(type="category")
    fig.show()
```

Variables à nombreuses valeurs (distributions)

```python
cols_to_plot = [c for c in X.columns if "grade" in c.lower() or "age" in c.lower()]
print("Variables tracées :", cols_to_plot)

for col in cols_to_plot:
    fig = px.histogram(df, x=col, color=df["Dropout"].map({0: "Non", 1: "Oui"}),
                       barmode="overlay", opacity=0.6,
                       title=f"Distribution de {col} selon le décrochage")
    fig.show()
```

Variables « codes » : taux de décrochage par modalité

```python
code_like = [c for c in X.columns if 3 <= X[c].nunique() <= 60]
print("Variables à plusieurs modalités :", code_like)

for col in code_like[:6]:   # on en regarde 6 pour commencer
    taux = df.groupby(col)["Dropout"].agg(["mean", "count"]).reset_index()
    fig = px.bar(taux, x=col, y="mean", hover_data=["count"],
                 title=f"Taux de décrochage par modalité de : {col}",
                 labels={"mean": "Taux de décrochage"})
    fig.update_xaxes(type="category")
    fig.show()
```

## Bilan des erreurs trouvées et corrections appliquées

| N° | Erreur / problème trouvé | Comment on l'a vu | Correction appliquée |
|---|---|---|---|
| 1 | Cible à 3 classes (Graduate, Dropout, Enrolled) | Répartition 49,9 % / 32,1 % / 17,9 % | Binarisation : `Dropout = 1`, les autres `= 0` (32,1 % de positifs) |
| 2 | Variables catégorielles codées en entiers (`Course`, `Application mode`, `Marital Status`, `Previous qualification`, `Nationality`, qualifications et métiers des parents) | `Course` va de 33 à 9991 : ce sont des codes, pas des quantités | Liste `cat_cols`, encodage one-hot dans le pipeline (`min_frequency=20`) |
| 3 | Faute dans un nom de colonne : `Nacionality` | Affichage de `X.columns` | Renommée en `Nationality` |
| 4 | Variable quasi constante sans effet : `Educational special needs` (98,8 % de la même valeur, taux de décrochage identique) | Cellule des valeurs dominantes + graphique par variable | Colonne supprimée |
| 5 | Variables quasi constantes `International` et `Nationality` (97,5 %) | Cellule des valeurs dominantes | **Conservées** : utiles pour l'analyse d'équité (variables sensibles) |
| 6 | Variables très corrélées entre elles (S1/S2 : 0,84 à 0,94 ; métier mère/père : 0,91) | Cellule des paires corrélées (> 0,8) | **Conservées** (modèles à arbres insensibles) ; régression logistique régularisée (`C=0.5`) ; option « avec / sans S2 » prévue dans l'app |
| 7 | Échelles très différentes entre variables numériques | `describe()` : notes sur 20, `Admission grade` ~127, taux économiques ~1 | `StandardScaler` sur les variables numériques |
| 8 | Classes déséquilibrées (68 % / 32 %) | Répartition de la cible | Séparation stratifiée + `class_weight="balanced"` ; métriques : AUC-ROC, PR-AUC, rappel, F1 (pas l'accuracy seule) |
| 9 | Risque de fuite de données (fit du préprocesseur sur tout le jeu) | Bonne pratique | `fit` du préprocesseur sur le train uniquement |
| 10 | Outliers détectés (surtout sur des variables binaires ou discrètes) | Cellule IQR | **Aucune action** : valeurs légitimes |
| 11 | Notes du 2e semestre très dominantes (corrélation -0,57) | Corrélation avec la cible | Option « avec / sans S2 » dans l'app (détection précoce vs tardive) |


```python
X2 = X.copy()
X2 = X2.rename(columns={"Nacionality": "Nationality"})

# Cible binaire
y2 = (y_raw == "Dropout").astype(int)
y2.name = "Dropout"

# Variable sans effet observé
X2 = X2.drop(columns=["Educational special needs"])

print(X2.shape, y2.mean().round(3))
```

```python
cat_cols = [
    "Marital Status", "Application mode", "Course",
    "Previous qualification", "Nationality",
    "Mother's qualification", "Father's qualification",
    "Mother's occupation", "Father's occupation",
]
bin_cols = [c for c in X2.columns
            if X2[c].nunique() == 2 and c not in cat_cols]
num_cols = [c for c in X2.columns if c not in cat_cols + bin_cols]

# Vérification : toutes les colonnes sont classées une seule fois
assert len(cat_cols) + len(bin_cols) + len(num_cols) == X2.shape[1]

print("Catégorielles :", len(cat_cols))
print("Binaires      :", len(bin_cols), bin_cols)
print("Numériques    :", len(num_cols))
```

```python
pd.DataFrame({
    "modalités": X2[cat_cols].nunique(),
    "modalités_rares_(<20)": [(X2[c].value_counts() < 20).sum() for c in cat_cols],
})
```

```python
from sklearn.model_selection import train_test_split
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder, StandardScaler

X_train, X_test, y_train, y_test = train_test_split(
    X2, y2, test_size=0.2, stratify=y2, random_state=42
)
print("Train :", X_train.shape, y_train.mean().round(3))
print("Test  :", X_test.shape, y_test.mean().round(3))

prep = ColumnTransformer([
    ("cat", OneHotEncoder(handle_unknown="ignore", min_frequency=20,
                          sparse_output=False), cat_cols),
    ("num", StandardScaler(), num_cols),
    ("bin", "passthrough", bin_cols),
])

Xt_train = prep.fit_transform(X_train)   # fit uniquement sur le train
Xt_test = prep.transform(X_test)
print("Après encodage :", Xt_train.shape)
```

```python
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import roc_auc_score, average_precision_score

lr = LogisticRegression(max_iter=2000, class_weight="balanced", C=0.5)
lr.fit(Xt_train, y_train)
p = lr.predict_proba(Xt_test)[:, 1]
print("AUC-ROC :", round(roc_auc_score(y_test, p), 3))
print("PR-AUC  :", round(average_precision_score(y_test, p), 3))
```
